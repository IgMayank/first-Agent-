```python
from langchain_openai import ChatOpenAI
import os
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
llm = ChatOpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    model="qwen/qwen-2.5-7b-instruct")
```

```python
from langchain_core.messages import HumanMessage , SystemMessage
query =input("Think of something: ")
res = llm.invoke([
    SystemMessage(content="You are an Akinator-like game. You guess what the user is thinking by asking one question at a time. Only ask yes/no questions. Be interactive, smart, and narrow down possibilities step by step. Never give long explanations."),
    HumanMessage(content=query)
])
```
